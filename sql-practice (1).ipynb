1. Return a list of all orders with the following columns:

- Order ID
- Purchase timestamp
- Customer city
- Customer state

```python
%%sql
ATTACH DATABASE 'orders.db' AS orders;

SELECT orders.order_id, orders.order_purchase_timestamp, customers.customer_city, customers.customer_state
FROM
orders
LEFT JOIN
customers ON orders.customer_id = customers.customer_id
LIMIT 5
```

I used left join for this querry because focus is on orders, while other cells are additional data.

```python
%%sql
ATTACH DATABASE 'products.db' AS products;
ATTACH DATABASE 'order_items.db' AS order_items;

SELECT o.order_id, p.product_category_name, oi.seller_id, oi.price 
FROM
order_items oi
LEFT JOIN
orders o ON oi.order_id = o.order_id
LEFT JOIN
products p ON oi.product_id = p.product_id
LIMIT 5
```

Calculate the average payment value for each customer state. Display the states from highest average payment to lowest.

```python
%%sql

ATTACH DATABASE 'order_payments.db' AS order_payments;

SELECT 
customer_state,
AVG(payment_value) AS avg_payment
FROM (
    SELECT
    o.order_id AS order_id,
    c.customer_state AS customer_state,
    op.payment_value AS payment_value
    FROM
    orders o
    JOIN customers c ON o.customer_id = c.customer_id
    JOIN order_payments op ON o.order_id = op.order_id 
)
GROUP BY customer_state
ORDER BY avg_payment DESC
```

Classify every payment into one of three categories:

- Low
- Medium
- High

Return:

- Order ID
- Payment value
- Payment category

```python
%%sql
--First lets check what is the range of payment
WITH ranked_data AS(
    SELECT
    MAX(payment_value) OVER() AS max_payment,
    MIN(payment_value) OVER () AS min_payment, 
    AVG(payment_value) OVER () AS avg_payment,
    payment_value,
    NTILE(100) OVER (ORDER BY payment_value) AS percentile
    FROM order_payments
    )
SELECT MAX(max_payment) AS max_p,
MIN(min_payment) OVER () AS min_p, 
AVG(avg_payment) OVER () AS avg_p,
payment_value,
percentile
FROM ranked_data
-WHERE (percentile = 25 OR percentile = 75 OR percentile = 100)
```

7. Find all customers whose total spending is greater than the average spending of all customers.

Return:

- Customer ID
- Total amount spent

```python
%%sql 
WITH customer_spending AS (
    SELECT
        c.customer_id,
        SUM(op.payment_value) AS total_spending
    FROM customers c
    JOIN orders o
        ON c.customer_id = o.customer_id
    JOIN order_payments op
        ON o.order_id = op.order_id
    GROUP BY c.customer_id
)

SELECT
    customer_id,
    total_spending
FROM customer_spending
WHERE total_spending >
(
    SELECT AVG(total_spending)
    FROM customer_spending
)
ORDER BY total_spending DESC

LIMIT 5
```

- Daily sales
- Cumulative sales

```python
%%sql
SELECT date,daily_sales, SUM(daily_sales) OVER(ORDER BY date) AS cumulative_sales
FROM
(SELECT
    DATE(o.order_purchase_timestamp) AS date, SUM(op.payment_value) AS daily_sales
    FROM
    orders o JOIN order_payments op ON o.order_id = op.order_id
GROUP BY date
ORDER BY date
)

LIMIT 10
```

12.

For every customer, calculate the number of days between consecutive purchases.

Return:

- Customer ID
- Order ID
- Purchase timestamp
- Days since previous purchase

```python
%%sql
-- noticed here for customer we should use customer_unique_id, not customer_id I was using in previous solutions
WITH customer_purchase_days AS (
    SELECT c.customer_unique_id AS customer_unique_id, 
    o.order_id AS order_id, 
    o.order_purchase_timestamp AS order_purchase_timestamp, 
    julianday(o.order_purchase_timestamp) AS current_day,
    julianday(LAG(o.order_purchase_timestamp) OVER(PARTITION BY c.customer_unique_id ORDER BY julianday(o.order_purchase_timestamp))) AS previous_day
    FROM
    orders o JOIN customers c ON o.customer_id = c.customer_id
)
SELECT customer_unique_id, 
    order_id, 
    order_purchase_timestamp,
    CAST((current_day - previous_day) AS INT) AS day_since_previous_purchase
FROM customer_purchase_days
WHERE day_since_previous_purchase NOT NULL AND day_since_previous_purchase <> 0
ORDER BY day_since_previous_purchase ASC

LIMIT 5
```

Find the three highest-spending customers in each state.

Return:

- State
- Customer ID
- Total spending

```python
%%sql
WITH customer_per_state AS (
SELECT c.customer_unique_id AS customer_unique_id,
        c.customer_state AS customer_state, 
        SUM(oi.price) AS total_price
FROM
    customers c JOIN orders o ON c.customer_id = o.customer_id JOIN order_items oi ON oi.order_id = o.order_id
GROUP BY c.customer_state, c.customer_unique_id
ORDER BY c.customer_state, c.customer_unique_id
)
SELECT customer_unique_id,
customer_state,
total_price
FROM (SELECT customer_unique_id,
        customer_state,
        total_price,
    ROW_NUMBER() OVER(PARTITION BY customer_state ORDER BY total_price DESC) AS row_num
    FROM    
    customer_per_state)
WHERE row_num IN (1,2,3)

LIMIT 5
```

14.

For each customer, find the most expensive product they have purchased.

Return:

- Customer ID
- Product ID
- Product category
- Price

```python
%%sql
--ATTACH DATABASE 'product_category_name.db' AS product_category_name;
WITH products_per_customer AS (
    SELECT
        c.customer_unique_id AS customer_unique_id,
        p.product_id AS product_id,
        oi.price AS price,
        pcn.product_category_name_english AS product_category_name_english,
        ROW_NUMBER() OVER(PARTITION BY c.customer_unique_id ORDER BY oi.price DESC) AS row_num
    FROM
    orders o JOIN customers c ON c.customer_id = o.customer_id
    JOIN order_items oi ON o.order_id = oi.order_id
    JOIN products p ON oi.product_id = p.product_id
    JOIN product_category_name pcn ON p.product_category_name = pcn.product_category_name
)
SELECT 
    customer_unique_id, product_id, price, product_category_name_english
FROM products_per_customer
WHERE row_num = 1
LIMIT 5
```

15.

Create a business summary for every state containing:

- State
- Number of unique customers
- Number of orders
- Total revenue
- Average order value
- Highest payment value

Sort the results by total revenue in descending order.

```python
%%sql
WITH order_summary AS (
    SELECT
        c.customer_state,
        c.customer_unique_id,
        o.order_id,
        SUM(oi.price) AS order_value,
        MAX(op.payment_value) AS highest_payment_value
    FROM customers c
    JOIN orders o 
        ON c.customer_id = o.customer_id
    JOIN order_items oi 
        ON o.order_id = oi.order_id
    JOIN order_payments op 
        ON o.order_id = op.order_id
    GROUP BY
        c.customer_state,
        c.customer_unique_id,
        o.order_id
)

SELECT
    customer_state,
    COUNT(DISTINCT customer_unique_id) AS customer_count,
    COUNT(order_id) AS order_count,
    SUM(order_value) AS total_revenue,
    AVG(order_value) AS average_order_value,
    MAX(highest_payment_value) AS highest_payment_value
FROM order_summary
GROUP BY customer_state
ORDER BY total_revenue DESC;
```
